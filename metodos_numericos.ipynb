```python
##EJERCICIO 1##
# Definir el punto de expansión de la serie de Taylor
x0 = 1  # Centro de la expansión
n = 6  # Serie de Taylor de orden 4

# Definir la variable simbólica
x = sp.Symbol('x')

# Obtener la serie de Taylor de ln(x) centrada en x0 con 5 términos
t = sp.ln(x).series(x, x0, 5).removeO()
print(t)  # Imprimir la serie de Taylor

# Convertir la expresión simbólica en una función numérica usable con numpy
t = sp.lambdify(x, t, 'numpy')

# Generar valores de x para la gráfica
# Ajustar el rango para evitar logaritmos de números no positivos
xx = np.linspace(0.1, x0 + 3, 10)  # Rango de valores de x, asegurando x > 0

# Calcular los valores reales de ln(x) y su aproximación con la serie de Taylor
yc = np.log(xx)  # Valores exactos de ln(x)
yt = t(xx)  # Valores aproximados con la serie de Taylor

# Graficar la función original y su aproximación
plt.plot(xx, yc, label="ln(x)")  # Gráfica de la función original
plt.plot(xx, yt, label="Serie de Taylor orden 5")  # Gráfica de la aproximación

# Agregar leyenda y título
plt.legend()
plt.title("Serie de Taylor de $f(x) = \\ln(x)$")

# Configuración adicional
plt.grid(False)  # Desactivar la cuadrícula

# Guardar la figura (opcional, descomentarlo si se necesita guardar)
# plt.savefig('seriet.pdf', bbox_inches='tight')

# Mostrar la gráfica
plt.show()
```

```python
##EJERCICIO 2##

# Importar las librerías
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# Definir el punto de expansión de la serie de Taylor
x0 = 0       # Centro de la expansión
n = 5        # Serie de Taylor de orden 4

# Definir la variable simbólica
x = sp.Symbol('x')

# Obtener la serie de Taylor de e^x
# centrada en x0 con n términos
t = sp.exp(x).series(x, x0, n).removeO()

print("Serie de Taylor:")
print(t)

# Convertir la expresión simbólica en una función
# numérica usable con numpy
t = sp.lambdify(x, t, 'numpy')

# Generar valores de x para la gráfica
xx = np.linspace(x0 - 3, x0 + 3, 10)

# Calcular los valores reales de e^x y su
# aproximación con la serie de Taylor
yc = np.exp(xx)      # Valores exactos de e^x
yt = t(xx)           # Valores aproximados con la serie

# Graficar la función original y su aproximación
plt.plot(xx, yc, label="e^x")
plt.plot(xx, yt, label="Serie de Taylor orden 4")

# Agregar leyenda y título
plt.legend()
plt.title("Serie de Taylor de f(x) = e^x")

# Configuración adicional
plt.grid(False)

# Mostrar la gráfica
plt.show()
```
